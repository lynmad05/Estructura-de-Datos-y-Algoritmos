```python

class Cola:
  def __init__(self,L=10):
    self.L= L
    self.A= []
    self.tope= 0

  def estaVacia(self):
    if self.tope==0: return True
    else:return False

  def estaLlena(self):
    if self.tope==self.L:
      return True
    else:
      return False 

  def encolar(self,dato):
    if not self.estaLlena():
      self.A.append(dato)
      self.tope+=1
    else:
      print("Cola llena")

  def desencolar(self):
    dato = None
    if not self.estaVacia():
      dato= self.A[0]
      self.tope-=1
      del self.A[0]
    else:
      print("Cola vacia")
    return dato

  def dividir_cola(self,c1,c2):
        cx=Cola(self.L)
        while not self.estaVacia():
            dato=self.desencolar()
            cx.encolar(dato)
            if dato%2==0:
                c1.encolar(dato)
            else:
                c2.encolar(dato)
        while not cx.estaVacia():
            self.encolar(cx.desencolar())


  def procesar_cola(self):
    cx=Cola(self.L)
    s=0
    m=0
    tope=0
    while not self.estaVacia():
          dato=self.desencolar()
          s+=dato
          if dato>m:
              m=dato
          tope+=1
    while not cx.estaVacia():
            self.encolar(cx.desencolar())
    print("suma de datos: ", s)
    print("Máximo de la cola:  ", m)
    print("Cantidad de elementos: ", tope)

  def mostrar(self):
    print(self.A)
   
```

```python
#ejercicio3
class Cola_original:
  def __init__(self,L=5):
    self.L= L
    self.A= [] 
    self.tope= 0 

  def estaVacia(self):
    if self.tope==0:
      return True
    else:
      return False

  def estaLlena(self):
    if self.tope==self.L:
      return True
    else:
      return False

  def encolar(self,dato):
    if not self.estaLlena():
      self.A.append(dato)
      self.tope+=1
    else:
      print("Cola llena")

  def desencolar(self):
    dato = None
    if not self.estaVacia():
      dato= self.A[0]
      self.tope-=1
      del self.A[0]
    else:
      print("Cola vacia")
    return dato

  def invertir_cola(self,c):
       if c.estaVacia():
         return
       dato=c.desencolar()
       self.invertir_cola(c)
       self.encolar(dato)
       
  def mostrar(self):
    print(self.A)
```

```python
# Crear la cola original
C = Cola_original()
C.encolar(1)
C.encolar(2)
C.encolar(3)
C.encolar(4)
C.encolar(5)


print("Cola original: ")
C.mostrar()

C.invertir_cola(C)
print("\nCola invertida:")
C.mostrar()
```
