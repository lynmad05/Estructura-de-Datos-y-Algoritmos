```python
class nodoD: 
    def __init__(self, dato):
        self.dato=dato
        self.izq=None
        self.der=None
    def __str__(self):
        return str(self.dato)+(' < - >')
   
    
```

```python
class DLL:  
    def __init__(self):  
        self.I = None  
        self.D = None  
  
    def prepend(self, dato):  
        aux = nodoD(dato)  
        if self.I == self.D and self.I == None:  
            self.I = self.D = aux  
        else:  
            aux.der = self.I
            if self.I is not None:  
                self.I.izq = aux  
            self.I = aux 
    #Append para insertar de der a izq
    def append(self, dato):  
        aux = nodoD(dato)  
        if self.I == self.D and self.I == None:  
            self.I = self.D = aux  
        else:  
            aux.izq = self.D  
            if self.D is not None:
                self.D.der = aux  
            self.D = aux
    
    #eliminar pop elimina de der a izq
    def pop(self):  
        if self.D is None:  
            return  
        if self.I == self.D:  
            self.I = self.D = None  
        else:  
            self.D = self.D.izq  
            self.D.der = None  
    #eliminar shift elimina de izq a der
    def shift(self):  
        if self.I is None:  
            return  
        if self.D == self.I:  
            self.D = self.I = None  
        else:  
            self.I = self.I.der 
            self.I.izq = None  
        
    #Get obtenga elementos 
    def get_valor(self,index):
        temp = self.I
        for i in range(index):
            if temp is None:
                return None
            temp = temp.der
        return temp.dato if temp else None
    
    #Update un elemento por el indice
    def update(self, index, new):
        temp = self.I
        for i in range(index):
            if temp is None:
                return
            temp = temp.der
        if temp:
            temp.dato = new
        
    #Insertar un elemento por posicion:
    def insertar(self, index, dato):  
        if index == 0:  
            self.prepend(dato)  
            return  
        new_node = nodoD(dato)  
        temp = self.I  
        for i in range(index - 1):  
            if temp is None:  
                return  
            temp = temp.der  
        if temp is None:  
            self.append(dato)  
        else:  
            new_node.der = temp.der  
            new_node.izq = temp  
            if temp.der:  
                temp.der.izq = new_node  
            temp.der = new_node  
            
    #Eliminar por posicion
    def eliminar(self, index):  
        if self.I is None:  
            return
          
        temp = self.I  
        for i in range(index):  
            if temp is None: 
                return  
            temp = temp.der  
        if temp is None:  
            return  
        if temp == self.I:  
            self.I = temp.der  
        if temp == self.D:  
            self.D = temp.izq  
        if temp.izq:  
            temp.izq.der = temp.der  
        if temp.der:  
            temp.der.izq = temp.izq  
        
            
    #Algoritmo que invierte la lista
    def reverse(self):  
        aux = self.I  
        temp = None  
        self.D = aux
        while aux is not None:  
            temp = aux.izq  
            aux.izq = aux.der  
            aux.der = temp  
            aux = aux.izq  
        if temp is not None:  
            self.I = temp.izq 
 
    def ImIaD(self):  
        print("Null",  end=' ')
        x = self.I  
        while x is not None:  
            print(x, end=' ')  
            x = x.der  
        print("Null")  
    
    def ImDaI(self): 
        print("Null",  end=' ') 
        x = self.D  
        while x is not None:  
            print(x, end=' ')  
            x = x.izq  
        print("Null")

```
