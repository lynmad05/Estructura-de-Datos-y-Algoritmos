```python
class nodo:
  def __init__(self,dato):
    self.dato =dato
    self.sgte = None

  def __str__(self):
    return str(self.dato) + " ->"
```

```python
class listaS:
  def __init__(self):
    self.Ini=None
    self.Fin =None

  def insertar(self,dato):
    aux=nodo(dato)
    if self.Ini==None:
      self.Ini=self.Fin=aux
      self.Fin.sgte=self.Ini
    else:
      self.Fin.sgte=aux
      self.Fin=aux
      self.Fin.sgte=self.Ini


  def eliminar_mul(self, k):
    if self.Ini is None:
        return
      
    while self.Ini.dato % k == 0: 
        if self.Ini == self.Fin:  # Si solo queda un nodo
            self.Ini = self.Fin = None # se elimina 
            return
        self.Ini = self.Ini.sgte # Aqui hacemos que cumpla el ciclo circular
        self.Fin.sgte = self.Ini  
      
    x = self.Ini
    while x.sgte != self.Ini:
        if x.sgte.dato % k == 0 :
            x.sgte = x.sgte.sgte # Salta el nodo que se elimina
        x = x.sgte  
    


  def imprimir(self,v ):
      if self.Ini is None:
        print("Lista vacía")
        return
      x=self.Ini
      c=0
      while c!=v:
        print(x.dato , end=" -> ")
        x=x.sgte
        if x==self.Ini:
          c+=1

      print("null")
      print('vueltas: ', c)
```

```python
L = listaS()
L.insertar(1)
L.insertar(2)
L.insertar(3)
L.insertar(4)
L.insertar(5)
L.insertar(6)
L.insertar(7)
L.insertar(8)
L.insertar(9)
L.insertar(10)
L.insertar(11)
L.insertar(12)
L.imprimir(1)

L.eliminar_mul(2)
L.imprimir(1)

```
